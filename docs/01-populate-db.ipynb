# Populate Database

This notebook demonstrates the pyeed ingestion API for populating the database with protein data from UniProt, enriched with Rhea reactions and ChEBI molecules.

The API provides two interfaces:
- **Functional API**: Individual async functions for fine-grained control
- **Fluent API**: Chainable Ingester class for concise workflows

## Setup

```python
from pyeed import (
    Database,
    Ingester,
    fetch_proteins_by_ids,
    fetch_proteins_by_interpro,
    enrich_reactions,
    enrich_molecules,
    ingest_full_pipeline,
)

from pyeed.ingest.embed import embed_proteins
from pyeed.embedding.pooling import mean_pooling

db = Database()
```

```python
await embed_proteins(
    db,
    pooling_methods=[mean_pooling],
    batch_size=64,
    chunk_size=1000,
)
```

## 1. Fluent API Examples

The Ingester class provides a clean, chainable interface for configuring pipelines.


### Simple: Fetch proteins by InterPro ID

```python
# Fetch proteins from InterPro family
stats = await Ingester(db).from_interpro("IPR002133").run()
print(f"Ingested {stats['proteins']} proteins")
```

### With Rhea reaction enrichment


```python
# Fetch proteins and enrich with Rhea reactions
stats = await Ingester(db).from_interpro("IPR002133").with_reactions().run()
print(f"Proteins: {stats['proteins']}, Reactions: {stats['reactions']}")
```

### Full enrichment (Rhea + ChEBI)


```python
# Full pipeline: proteins + reactions + molecules
stats = await Ingester(db).from_interpro("IPR002133").with_full_enrichment().run()
print(
    f"Proteins: {stats['proteins']}, Reactions: {stats['reactions']}, Molecules: {stats['molecules']}"
)
```

### Fetch by specific UniProt IDs


```python
# Fetch specific proteins by accession IDs
stats = await Ingester(db).from_ids(["P12345", "Q9Y6K9", "P00374"]).with_full_enrichment().run()
print(f"Ingested {stats['proteins']} proteins with full enrichment")
```

## 2. Functional API Examples

For more control, use individual async functions.


### Step-by-step ingestion


```python
# Fetch proteins
# proteins = await fetch_proteins_by_interpro(db, "IPR002133")
# print(f"Fetched {len(proteins)} proteins")

# Enrich with reactions
reaction_count = await enrich_reactions(db)
print(f"Enriched {reaction_count} reactions")

# Enrich with molecules
molecule_count = await enrich_molecules(db)
print(f"Enriched {molecule_count} molecules")
```

### All-in-one pipeline


```python
# Single function call for the entire pipeline
stats = await ingest_full_pipeline(
    db,
    interpro_id="IPR002133",
    include_reactions=True,
    include_molecules=True,
)
print(f"Pipeline complete: {stats}")
```

## 3. Advanced Options


### Quiet mode (no progress bars)


```python
# Disable progress bars (useful for CI/scripts)
stats = await Ingester(db).from_interpro("IPR002133").quiet().run()
print(f"Completed quietly: {stats}")
```

### Custom batch sizes


```python
# Tune performance with custom batch sizes
stats = await (
    Ingester(db).from_ids(["P12345", "Q9Y6K9"]).with_batch_size(100).with_chunk_size(50).run()
)
print(f"Completed with custom settings: {stats}")
```

## Summary

The pyeed ingestion API offers:

- **Fluent API**: Clean, chainable syntax with `Ingester(db).from_*().with_*().run()`
- **Functional API**: Individual async functions for fine-grained control
- **All-in-one**: Single `ingest_full_pipeline()` call for complete workflows
- **Flexible enrichment**: Optionally include Rhea reactions and ChEBI molecules
- **Performance tuning**: Configure batch sizes, chunk sizes, and concurrency
- **Progress tracking**: Rich progress bars (can be disabled for scripts)

All functions automatically ensure schema is synchronized (can be skipped with `skip_schema_sync=True`).
